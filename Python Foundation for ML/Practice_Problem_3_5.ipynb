Problem - 01


```python
def create_file(filename, content):
    with open(f"{filename}", 'w') as f:
      f.write(f"{content}")

create_file("note.txt", "Hello, this is my first file.")

```

Problem - 02


```python
def read_lines(filename):
    with open(f"{filename}", 'r') as f:
      print("1:",f.readline(), end = "")
      print("2:",f.readline(), end = "")
      print("3:",f.readline(), end = "")

read_lines("data.txt")
```

Problem - 03

```python
def add_log(message):
    with open("log.txt", 'a') as f:
      f.write(f"{message}\n")

add_log("User logged in")
add_log("File uploaded")
```

Problem - 04

```python
def read_from_position(filename, position):
  with open(f"{filename}", 'r') as f:
    f.seek(position)
    print(f.read())

read_from_position("hello.txt", 7)
```

Problem - 05

```python
def safe_read(filename):
    try:
      f = open(f"{filename}", 'r')

    except FileNotFoundError:
      print("Error: File not found.")

    else:
      print(f.read())
      f.close()

    finally:
      print("Operation complete.")

safe_read("missing.txt")
safe_read("hello.txt")
```
