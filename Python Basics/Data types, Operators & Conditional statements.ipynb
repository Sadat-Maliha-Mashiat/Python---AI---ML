# Conditional Statements

```python
a=int(input("Enter the Number:"))
if (a%2==0):
  print("Even")
else: print("ODD")
```

```python
a=int(input("Enter the Number:"))
if a>0:
  print("Positive")
elif a<0:
  print("Negative")
else:
  print("Zero")
```

```python
a,b=map(int,input("Enter Values:").split())
ch=input("Enter Operator:")

if ch=='+':
  print(a+b)
elif ch=='-':
  print(a-b)
elif ch=='*':
  print(a*b)
else:
 print(a/b)
```

```python
vowel="aeiou"
ch=input("Enter the character:")
if ch in vowel:
  print("Even")
else:
  print("ODD")
```
