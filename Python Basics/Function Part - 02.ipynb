# Lambda Function
- anonymous function/unnamed function
- automatically returns a value
- Syntax: lambda parameter:expression

```python
lst=2

sq=lambda x:x*x

print(sq(lst))
```

```python
a,b,c=3,4,5

avg=lambda x,y,z:(x+y+z)/3

print(avg(a,b,c))
```

```python
lst=[1,2,3,4,5]

tsum=lambda x:sum(x)

print(tsum(lst))
```

```python
a=55

ch=lambda x:"Even" if x%2==0 else "Odd"

print(ch(a))
```

# comprehension

- used when we want to create a new list,set,dictionary from an existing iterator.

- List Comprehension: [expression for item in list_name]

- Set Comprehension: {expression for item in set_name}

- Dictionary Comprehension: {key:value for item in iterator_name}

```python
lst=[1,2,3,4,5]

nlst=[]

for i in lst:
  nlst.append(i**2)

print(nlst)
```

```python
#comprehension

lst=[1,2,3,4,5]

nlst=[x**2 for x in lst]

print(nlst)
```

```python
lst=[1,2,3,4,5]

lsq=lambda lt:[x*x for x in lt]

print(lsq(lst))
```

```python
# Sort

num=[3,2,4,5,1,6]

print(sorted(num))
```

```python
st=[("A",85,56),("C",92,45),("B",78,100)]

result=sorted(st,key=lambda x:x[2])
print(result)
```

```python
st=[("A",85,56),("C",92,45),("B",78,100)]

result=sorted(st,key=lambda x:x[2],reverse=True)
print(result)
```

```python
st={"A":85,"C":92,"B":78}

result=sorted(st.items(),key=lambda x:x[1])
print(result)
```

```python
st={"A":85,"C":92,"B":78}

result=max(st.items(),key=lambda x:x[1])
print(result)
```

```python
st={"A":[85,67],"C":[92,56],"B":[78,58]}

result=sorted(st.items(),key=lambda x:x[1][1])
print(result)
```

# map

- syntax : map(function_expression,parameters)

```python
lst=[11,12,13,14,15]

result=list(map(lambda x:x**2,lst))

print(result)
```
