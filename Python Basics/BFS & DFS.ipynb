BFS(graph, start_node):

    create empty set visited
    create queue and enqueue start_node

    while queue is not empty:
        current_node = dequeue from queue

        if current_node not in visited:
            visit current_node (e.g., print or record it)
            add current_node to visited

            for each neighbor in graph[current_node]:
                if neighbor not in visited:
                    enqueue neighbor

```python
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['A', 'E', 'F'],
    'C': ['A', 'F'],
    'D': ['A'],
    'E': ['B'],
    'F': ['B', 'C']
}

start_node = 'A'
```

```python
def bfs(graph, start_node):
  visited = []
  queue = []

  queue.append(start_node)

  while queue:
    current_node = queue.pop(0)

    if current_node not in visited:
      visited.append(current_node)
      print(current_node)

      for node in graph[current_node]:
        if node not in visited:
          queue.append(node)
```

```python
bfs(graph,start_node)
```

DFS_Iterative(graph, start_node):
    
    create empty set visited
    
    create stack and push start_node

    while stack is not empty:
        current_node = pop from stack

        if current_node not in visited:
            visit current_node (e.g., print or record it)
            add current_node to visited

            for each neighbor in reverse(graph[current_node]):
                if neighbor not in visited:
                    push neighbor onto stack


```python
graph = {
    'A' : ['B', 'C', 'D'],
    'B' : ['A', 'D'],
    'C' : ['A', 'D', 'E'],
    'D' : ['A', 'B', 'C', 'E'],
    'E' : ['C', 'D']
}

start_node = 'A'
```

```python
def dfs_iter(graph, start_node):
  visited = []
  stack = []

  stack.append(start_node)

  while len(stack) != 0:
    current_node = stack.pop()

    if current_node not in visited:
      print(current_node)
      visited.append(current_node)

      for cnn in graph[current_node]:
        if cnn not in visited:
          stack.append(cnn)
```

```python
dfs_iter(graph, start_node)
```
