## 🧮 Python Variables and Data Types

### 🧱 Integer
Whole numbers (positive or negative) without decimals.

```python
x = 10
y = -25

print(x, y)
print(type(x), type(y))

# Different printing styles
print(x, y, sep=",")
print(x, ",", y)
print(x, ',', y)
print(str(x) + ", " + str(y))
```

### 🧠 String with Placeholders (.format and f-string)
Different methods to print formatted strings.

```python
x, y = 10, -25

# Using .format()
print("X = {}, Y = {}".format(x, y))
print("Name: {name}, Age: {age}".format(name="Rahim", age=22))

# Floating-point formatting
pi = 3.14159265
print("Pi = {:.2f}".format(pi))  # 2 decimal places

# Using f-string
x, y = 1, 1.5
print(f"X = {x}, Y = {y}")

name = "Rahim"
age = 22
print(f"My name is {name} and I am {age} years old.")

a, b = 5, 3
print(f"{a} + {b} = {a+b}")

pi = 3.14159265
print(f"Pi rounded to 2 decimals: {pi:.2f}")
```

### 🌊 Float
Numbers with decimal points.

```python
a,b,c,d=input("Enter the values").split()
a,b,c,d=float(a),float(b),float(c),float(d)
print(f"Result of {c}*{d} - {a}+{b} is = {c*d-a+b:.2f}")
```

```python
pi = 3.1416
temp = -12.5

print(pi, temp)
print(type(pi), type(temp))
```

### 🧵 String
Text enclosed in quotes.

```python
name = "Mashiat"
greeting = 'Hello'
multiline = """This is
a multiline
string."""

print(name)
print(greeting)
print(multiline)
print(type(name))
```

### ⚙️ Boolean
True or False values.

```python
is_student = True
is_graduated = False

print(is_student, is_graduated)
print(type(is_student))
```

## 📥 Python Input and Output

### 🖨️ Output (print)

```python
# 1. Printing simple values
print(10)
print("Hello Python")

# 2. Printing multiple values
x, y = 5, 10
print(x, y)

# 3. Custom separator using sep
print(x, y, sep=", ")

# 4. Formatted output using .format()
print("X = {}, Y = {}".format(x, y))

# 5. Formatted output using f-strings
print(f"X = {x}, Y = {y}")

# 6. Concatenation with strings
print("X = " + str(x) + ", Y = " + str(y))
```

### 🧾 Multi-line Strings

```python
s = '''
Today we learned:
1. Output
2. Input
3. Data Types
4. If-Else
'''
print(s)
```

### 💡 Input (input)

```python
# 1. Simple input
name = input("Enter your name: ")
print("Hello,", name)

# 2. Input numbers
age = int(input("Enter your age: "))
print("Your age is", age)

# 3. Multiple inputs in one line
x, y = map(int, input("Enter two numbers separated by space: ").split())
print(f"Sum = {x + y}")
```

## 🔠 The split() Method
Break a string into parts based on a separator.

```python
# Split by comma
s = "apple,banana,cherry"
fruits = s.split(",")
print(fruits)

# Split by hyphen
s = "2025-09-26"
date_parts = s.split("-")
print(date_parts)

# Split with maxsplit
s = "one two three four"
parts = s.split(" ", 2)
print(parts)

# Using split() in input
numbers = input("Enter numbers separated by comma: ").split(",")
print(numbers)
```

### 🔄 split() vs sep in print()

| Function | Purpose | Direction | Default | Example |
|-----------|----------|------------|----------|----------|
| `split()` | Break string into list | Input → list | Whitespace | `"a b c".split()` → `['a','b','c']` |
| `sep` in `print()` | Join items with separator | Items → string | Space `" "` | `print(1,2,3, sep=",")` → `1,2,3` |


```python
x, y, z = 5, 10, 15

# Default separator
print(x, y, z)

# Custom separator
print(x, y, z, sep=", ")
print(x, y, z, sep=" | ")

# Using sep with different types
name = "Rahim"
age = 22
print("Name:", name, "Age:", age, sep=" -> ")

# Combining sep and end
print("Hello", "Python", sep="*", end="!!!\n")
print("Next line")
```

```python
# Type conversion
x = int(input("Enter an integer: "))
y = float(input("Enter a float: "))
print(f"Integer: {x}, Float: {y}")

# Multiple values
a, b, c = map(int, input("Enter three numbers separated by space: ").split())
print(f"Sum = {a+b+c}")

# Input as list
numbers = list(map(int, input("Enter numbers separated by space: ").split()))
print(numbers)

# Input validation
num = input("Enter a positive number: ")
if num.isdigit() and int(num) > 0:
    print("Valid input:", num)
else:
    print("Invalid input")
```
