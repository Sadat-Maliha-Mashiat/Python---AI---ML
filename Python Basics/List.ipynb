Adding Elements into a list

- append() : add one element at the end of the list
- extend() : add two lists
- insert() : insert a value at a specific index

```python
lst=[1,2,3,4,5]
print(lst)
lst.append(7)
print(lst)
lst.extend([6,7])
print(lst)
lst.insert(2,9)
print(lst)
```

Removing Elements

- remove() : removes a specific value
- pop() : pop out a value from an index
- clear() : clear a whole list

```python
lst=[11,12,13,14,15]
lst.remove(12)
print(lst)
temp=lst.pop(2)
print(temp)
print(lst)

lst.clear()
print(lst)
```

```python
# Finding Frequency for each unique element in a list
course=list(input("Enter the Name:").split())

visited=[]

for item in course:
  if item not in visited:
    print(f"{item}:{course.count(item)}")
    visited.append(item)
```

```python
# displaying only unique element in a list

lst=list(map(int,input("Enter values:").split()))
visited=[]

for item in lst:
  if item not in visited:
    visited.append(item)
print(visited)
```
