Problem - 02:

```python
List = [1, 2, 3, 4, 5, 6, 7, 8]

def batch_generator(chunk_size, List):
    for i in range(0, len(List), chunk_size):
        yield List[i : i + chunk_size]

x = batch_generator(3, List)
print(next(x))
print(next(x))
print(next(x))
```

Problem - 04:

```python
def power(base, exponent = 2):
    return (base**exponent)

print(power(5))
print(power(5, 3))
```
