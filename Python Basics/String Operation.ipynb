## 1. Creating Strings
Strings are sequences of characters enclosed in single `' '` or double `" "` quotes.

```python
# Single and double quotes
s1 = 'Hello'
s2 = "World"
print(s1, s2)

# Multi-line string
s3 = """This is
a multi-line
string"""
print(s3)
```

## 2. Accessing Characters
Strings are indexed; access characters using `[]`.

```python
s = "Python"
print(s[0])   # First character
print(s[-1])  # Last character
```

## 3. Slicing Strings
Extract substring using `[start:end:step]`.

```python
s = "Python"
print(s[0:4])
print(s[2:])
print(s[:4])
print(s[::2])
```

## 4. String Length
Use `len()` to get the number of characters.

```python
s = "Python"
print(len(s))  # 6
```

## 5. String Concatenation
Combine strings using `+`.

```python
s1 = "Hello"
s2 = "World"
s3 = s1 + " " + s2
print(s3)  # 'Hello World'
```

## 6. String Repetition
Repeat strings using `*`.

```python
s = "Hi! "
print(s * 3)  # 'Hi! Hi! Hi! '
```

## 7. String Membership
Use `in` and `not in` to check substring presence.

```python
s = "Python"
print("Py" in s)    # True
print("Java" not in s)  # True
```

## 8. Changing Case
Use `upper()`, `lower()`, `title()`, `capitalize()`, `swapcase()`.

```python
s = "python Programming"
print(s.upper())      # 'PYTHON PROGRAMMING'
print(s.lower())      # 'python programming'
print(s.title())      # 'Python Programming'
print(s.capitalize()) # 'Python programming'
print(s.swapcase())   # 'PYTHON PROGRAMMING'
```

## 10. Finding & Replacing
Use `find()` to get index, `replace()` to replace substrings.

```python
s = "Python is fun"
print(s.find("is"))       # 7
print(s.replace("fun", "awesome"))  # 'Python is awesome'
```

## 12. String Formatting
Insert variables using f-strings or `format()`.

```python
name = "Alice"
age = 25

# f-string
print(f"My name is {name} and I am {age} years old.")

# format()
print("My name is {} and I am {} years old.".format(name, age))
```

## 13. Checking String Content
Use `isalpha()`, `isdigit()`, `isalnum()`, `isspace()`, `islower()`, `isupper()`.

```python
s1 = "Hello"
s2 = "123"
s3 = "Hello123"
s4 = " "

print(s1.isalpha())   # True
print(s2.isdigit())   # True
print(s3.isalnum())   # True
print(s4.isspace())   # True
print(s1.isupper())   # False
print(s1.islower())   # False
```

## 14. String Reversal
Reverse a string using slicing `[::-1]`.

```python
s = "Python"
reversed_s = s[::-1]
print(reversed_s)  # 'nohtyP'
```

## 17. String Comparison
Compare strings using `==`, `!=`, `<`, `>`, `<=`, `>=`.

```python
s1 = "apple"
s2 = "banana"
print(s1 == s2)   # False
print(s1 < s2)    # True (lexicographical order)
```
