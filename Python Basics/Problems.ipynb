1. Write a Python function that accepts six numbers as
input. Calculate the sum of the first two numbers, the
product of the third and fourth numbers, and the fifth
number raised to the power of the sixth number. Display
each result along with the final sum of these three
computed values.

```python
def calculate(num):
    if len(num) != 6:
        print("Please enter exactly 6 numbers.")
        return

    addition = num[0] + num[1]
    multiplication = num[2] * num[3]
    power = num[4] ** num[5]
    total = addition + multiplication + power

    print(f"Addition = {addition}")
    print(f"Multiplication = {multiplication}")
    print(f"Power = {power}")
    print(f"Final Result = {total}")

num = list(map(int, input("Input: ").split()))
calculate(num)
```

5. Write a Python program to calculate the current age
from the given date of birth.

```python
from datetime import date

day, month, year = map(int, input("Enter Date of Birth (DD/MM/YYYY): ").split('/'))

date_of_birth = date(year, month, day)
today = date.today()

age = today.year - date_of_birth.year

if (today.month < date_of_birth.month):
  age -= 1

elif (today.month == date_of_birth.month) and (today.day < date_of_birth.day):
  age -= 1

print("Age =", age)
```

8. Write a Python function that accepts an integer as input
and determines whether the number is prime.

```python
def check_prime(num):
    if num <= 1:
        print("Not a prime number")

    else:
        flag = True

        for i in range(2, num):
            if num % i == 0:
                flag = False
                break

        if flag:
            print("Prime number")

        else:
            print("Not a prime number")

num = int(input("Enter a number: "))
check_prime(num)
```

10. Write a Python function multiply_all_numbers(*args)
that accepts any number of numerical arguments and
returns their product. If no arguments are provided,
return 1.

```python
def multiply_all_numbers(*values):
    if len(values) == 0:
        return 1

    res = 1

    for num in values:
        res *= num
    return res

values = list(map(int, input("Enter numbers: ").split()))
result = multiply_all_numbers(*values)

print(f"Product = {result}")
```

13. Write a Python function that calculates the factorial of a
given non-negative integer.

```python
def fact(n):
    if n == 1:
        return 1
    else:
        return n * fact(n - 1)

num = int(input("Enter a number: "))

print(f"Factorial = {fact(num)}")
```

14. Write a Python function that accepts a positive integer n
and generates the first n terms of the Fibonacci series.

```python
def fibonacci(n):
    fib = [0, 1]

    for i in range(2, n):
        fib.append(fib[-1] + fib[-2])
    return fib[:n]

n = int(input("Enter a number: "))
result = fibonacci(n)

print("Series =", *result)
```

15. Write a Python function that accepts two positive
integers and computes their Greatest Common Divisor
(GCD) using the Euclidean algorithm.

```python
def gcd(a, b):
    if b == 0:
        return a

    else:
        return gcd(b, a % b)

a, b = map(int, input("Enter two numbers: ").split())

print(f"GCD = {gcd(a, b)}")
```

16. Write a Python function that accepts two positive
integers and computes their Least Common Multiple
(LCM).

```python
def gcd(a, b):
    if b == 0:
        return a

    else:
        return gcd(b, a % b)

def lcm(a, b):
    return (a * b) // gcd(a, b)

a, b = map(int, input("Enter two numbers: ").split())

print(f"LCM = {lcm(a, b)}")
```

17. Write a Python program to calculate the sum of all prime
numbers within a given range.

```python
start, end = map(int, input("Enter range: ").split())
prime_sum = 0

for num in range(start, end + 1):
    if num > 1:
        flag = True

        for i in range(2, num):
            if num % i == 0:
                flag = False
                break

        if flag:
            prime_sum += num

print(f"Sum = {prime_sum}")
```

18. Write a Python program to check whether a given
number is a palindrome or not.

```python
num = int(input("Enter a number: "))

temporary = num
reverse = 0

while temporary > 0:
    digit = temporary % 10
    reverse = reverse * 10 + digit
    temporary //= 10

if num == reverse:
    print("Palindrome")

else:
    print("Not Palindrome")
```

32. Write a Python program to merge multiple dictionaries.
If a key appears in more than one dictionary, add the
corresponding values.

```python
n1 = int(input("Enter the number of items in the first dictionary: "))

d1 = {}

for i in range(n1):
    key = input(f"Enter key {i+1} for first dictionary: ")
    value = int(input(f"Enter value for {key}: "))
    d1[key] = value

n2 = int(input("\nEnter the number of items in the second dictionary: "))

d2 = {}

for i in range(n2):
    key = input(f"Enter key {i+1} for second dictionary: ")
    value = int(input(f"Enter value for {key}: "))
    d2[key] = value

result = d1.copy()

for key, value in d2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print(result)
```

33. Write a Python program to analyze student course
enrollment records and determine the course with the
highest number of enrolled students.

```python
n = int(input("Enter the number of students: "))

students = {}

for i in range(n):
    name = input(f"\nEnter the name of student {i+1}: ")
    m = int(input(f"How many courses does {name} take? "))

    courses = []
    for j in range(m):
        course = input(f"Enter course {j+1}: ")
        courses.append(course)

    students[name] = courses

count = {}

for courses in students.values():
    for course in courses:
       count[course] = count.get(course, 0) + 1

result = max(count, key = count.get)
print("Course:", result)
print("Total Enrolled:", count[result])
```

40. Write a Python program to count the number of vowels,
consonants, digits, and special characters in a string.

```python
text = input("Enter a string: ")

vowel = 0
consonant = 0
digit = 0
special = 0

for ch in text:
    if ch in "aeiouAEIOU":
        vowel += 1

    elif ch.isalpha():
        consonant += 1

    elif ch.isdigit():
        digit += 1

    elif not ch.isalnum():
        special += 1

print("Vowels =", vowel)
print("Consonants =", consonant)
print("Digits =", digit)
print("Special =", special)
```

42. Write a Python program to validate a password based on
the following conditions: it must contain at least one
uppercase letter, at least one digit, one special character,
and no more than five lowercase letters.

```python
password = input("Enter password: ")

upper = 0
lower = 0
digit = 0
special = 0

for ch in password:
    if ch.isupper():
        upper += 1

    elif ch.islower():
        lower += 1

    elif ch.isdigit():
        digit += 1

    elif not ch.isalnum():
        special += 1

if upper >= 1 and digit >= 1 and special == 1 and lower <= 5:
    print("Valid Password")

else:
    print("Invalid Password")
```
