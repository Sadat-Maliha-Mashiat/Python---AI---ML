# Python Slicing Operations
Learn all types of slicing operations for sequences like lists, strings, and tuples.

## 1. Slicing Definition
Slicing means extracting a portion/subsequence from a sequence type — such as string, list, tuple, etc.

Syntax: `sequence[start:end:step]`

- start: index where slice begins (*inclusive*)
  - Default = 0 if step > 0
  - Default = len(sequence)-1 if step < 0
- end: index where slice ends (*exclusive*)
  - Default = len(sequence) if step > 0
  - Default = -1 if step < 0
- step: difference between each index (*optional*, default = 1)

## 2. Basic Forward Slicing

```python
lst = [10, 20, 30, 40, 50]
print(lst[1:4])
```

## 3. Omitting Start Index

```python
print(lst[:3])
```

## 4. Omitting End Index

```python
print(lst[2:])
```

## 5. Omitting Both Start and End

```python
print(lst[:])
```

## 6. Using Step

```python
print(lst[0:5:2])
```

## 7. Negative Indexing

```python
print(lst[-4:-1])
```

## 8. Negative Step (Reversing)

```python
print(lst[::-1])
```

## 9. Reverse From Last to Specific Index

```python
print(lst[-1:0:-1])
```

## 10. Reverse With Step > 1

```python

print(text[-1:0:-2])
```

## 11. Single Element Access vs Slice

```python
print(lst[-1])
print(lst[-1::1])
```

## 12. Slicing Strings

```python
text = "Python"
print(text[1:5])
print(text[::-1])
```

## 13. Slicing Tuples

```python
t = (1, 2, 3, 4, 5)
print(t[::2])
```

## 14. Out-of-Range Slicing

```python
print(lst[1:100])
```

## 15. Slicing Nested Lists

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
print(matrix[0:2])
print(matrix[1][1:])
```

## 16. Copying a Sequence

*list2 = list1[:]*
- Will not affect both list as they share ddifferent memory location
- Also known as Shallow Copy

```python
lst2 = lst[:]
lst2.append(60)
print(lst)
print(lst2)
```
