- Collection of key-value pairs.
- Mutable as can be modified after creation.
- Keys must be unique & Keys must be immutable data types.
- Values can be of any data type.



```python
#create a dictionary

st={"Name":"Rahim","id":10,100:300}
print(st)

st2=dict(name="Rahim",id=100)
print(st2)

#empty dictionary
d={}
print(type(d))
```

```python
# accessing values

st={"Name":"Rahim","id":10,100:300}

print(st["Name"])

print(st.get("Name"))
```

```python
# taking input from user

st={}
n=int(input("Enter no. of students:"))

for i in range(n):
  key=input(f"Enter Key for {i+1}th Student:")
  value=int(input(f"Enter value for {i+1}th Student:"))
  st[key]=value

print(st)
```

```python
# update dictionary

st={'Rahim': '100', 'Karim': '200', 'Mahin': '500'}

st["Rahim"]=500
st[500]="Ramisa"

print(st)
```

```python
#length
st={'Rahim': 500, 'Karim': '200', 'Mahin': '300', 500: 'Ramisa'}
print(len(st))
```

```python
#Traversal
st={'Rahim': 500, 'Karim': 200, 'Mahin': 300, 'Ramisa':800}

#Only access key
for k in st.keys():
  print(k)

#only access value
for v in st.values():
  print(v)

#access both
for k,v in st.items():
  print(k,v)

```

```python
#Deleting elements: del, pop(), popitem(), clear()

st={'Rahim': 500, 'Karim': 200, 'Mahin': 300, 'Ramisa':800}

del st["Rahim"]
print(st)

st.pop("Karim")
print(st)

st.popitem()
print(st)

st.clear()
print(st)
```

```python
# Membership: in, keys(),values(),items()
st={'Rahim': 500, 'Karim': 200, 'Mahin': 300, 'Ramisa':800}

print(500 in st)
```

```python
# Dictionary construction from Lists
keys=['a','b','c','d','e']
values=[1,2,3,4,5]

ndct=dict(zip(keys,values))

print(ndct)
```

```python
dct={'a': 5, 'b': 2, 'c': 3, 'd': 4}

key=[]
value=[]

for k,v in dct.items():
  key.append(k)
  value.append(v)

print(key)
print(value)
```

```python
dct={'a': 5, 'b': 2, 'c': 3, 'd': 4}

key=""
value=""

for k,v in dct.items():
  key+=k
  value+=str(v)

print(key)
print(value)
```

```python
# sort a dictionary

dct={'b': 5, 'c': 2, 'a': 3, 'd': 4}

# sort by keys in asecending order

sorted_dct_key=dict(sorted(dct.items()))

print(sorted_dct_key)

# sort by keys in descending order

sorted_dct_key=dict(sorted(dct.items(),reverse=True))

print(sorted_dct_key)

# sort by values in ascending order

sorted_dct_val_as=dict(sorted(dct.items(),key=lambda item: item[1]))

print(sorted_dct_val_as)

# sort by values in descending order

sorted_dct_val_ds=dict(sorted(dct.items(),key=lambda item: item[1],reverse=True))

print(sorted_dct_val_ds)

```

```python
dct={'b': 5, 'c': 2, 'a': 3, 'd': 4}

#max & min key
print(max(dct))
print(min(dct))

#max & min value
print(max(dct.values()))
print(min(dct.values()))
```
