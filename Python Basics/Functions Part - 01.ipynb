```python
def calculator(a,b):
  return a+b,a*b,a-b,a/b

w,x,y,z=calculator(12,3)

print(w)
print(x)
```

```python
# passing a list

def average(num):
  return sum(num)/len(num)

num=[1,2,3,4,5]
print(average(num))

# passing a tuple

def summation(num):
  return sum(num)

num=(1,2,3,4,5)
print(summation(num))
```

```python
# passing a dictionary

def display(nwdct):
  for k,v in nwdct.items():
    print(k,v)

dct={'b': 5, 'c': 2, 'a': 3, 'd': 4}
display(dct)
```

```python
def country(name="Bangladesh"):
  print(name)

country()
country("Canada")
```

```python
# variable length positional arguments (*args)->treats as tuple

def total(*a):
  print(sum(a))

total(10,4)
total(10,4,5)
total(10,4,5,6)
```

```python
# variable length keyword arguments (**kwargs)

def student(**info):
  print(info)

student(name="Rahim")
student(name="Rahim", age=22)
student(name="Rahim", age=22, cgpa=3.6)
```

```python
def student(**info):
  for k,v in info.items():
    print(k)

student(name="Rahim", age=22, cgpa=3.6)
```

```python
# combination of *args & **kwargs
def student(x,y,*a,**b):
  print(a)

student("Rahim",80,90,50,dept="CSE",Sem="9th")
```

```python
def fact(n):
  if n==1:
    return 1
  else:
     return n*fact(n-1)


res= fact(5)
print(res)
```

```python
#gcd

def gcd(a,b):
  if b==0:
    return a
  else:
    return gcd(b,a%b)

a,b=48,12

print(gcd(a,b))

#lCM
f=(a*b)/gcd(a,b)
print(f)
```
