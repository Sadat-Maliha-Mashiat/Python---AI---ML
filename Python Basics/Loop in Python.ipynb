# Loops in Python
Python provides different types of loops to execute a block of code multiple times. Loops are essential for repetition, automation, and iteration over sequences like lists, strings, and ranges.

---

## 1.1 For Loop
**Used for sequential iteration** over a sequence such as a list, tuple, string, or range.

**Syntax:**
```python
for variable in sequence:
    # statements
```

**Example:**

````python
for i in range(5):
    print(i)
````

**Output:**
```
0
1
2
3
4
```

````python
flag=1;
n=int(input("Enter a Number:"))
if n<=1:
  print("Not Prime")
  flag=0

for i in range(2,n):
  if n%i==0:
    flag=0
    break
if flag==1:
  print("Prime Number")
else:
  print("Not Prime")
````

## 1.2 While Loop
**Repeats a block of code as long as a condition is true.**

**Syntax:**
```python
while condition:
    # statements
```

**Example:**

````python
i = 1
while i <= 5:
    print(i)
    i += 1
````

**Output:**
```
1
2
3
4
5
```

## 1.3 Nested Loops
**A loop inside another loop.**

**Example:**

````python
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
````

**Output:**
```
i=0, j=0
i=0, j=1
i=1, j=0
i=1, j=1
i=2, j=0
i=2, j=1
```

## 1.4 Loop Control Statements
| Statement | Purpose |
|------------|--------------------------------------|
| `break`    | Exit the loop completely              |
| `continue` | Skip current iteration and continue next |
| `pass`     | Placeholder, does nothing             |

**Example:**

````python
for i in range(5):
    if i == 2:
        continue
    print(i)
````

**Output:**
```
0
1
3
4
```

## 2. Range Function in Python
Used to generate a sequence of numbers.

**Syntax:** `range(start, stop, step)`
- `start` → beginning value (default = 0)
- `stop` → end value (excluded)
- `step` → difference between numbers (default = 1)

````python
print(list(range(10)))          # [0, 1, 2, 3, 4]
print(list(range(2, 7)))       # [2, 3, 4, 5, 6]
print(list(range(2, 12, 2)))   # [2, 4, 6, 8, 10]
print(list(range(0, -2, -1)))   # [5, 4, 3]
````

## 3. Applying Conditions in For Loops

````python
sum,prod=0,1
for i in range(3, 50):
  if i%2==0 and i%5==0:
    sum+=i
  if i%3==0 and i%7==0:
    prod*=i
print(f"Sum is:{sum}\nProduct is:{prod}")
````

**Output:** `6 12 18`
