## Arithmetic Operators

| Operator | Description | Example | Output |
|-----------|-------------|----------|---------|
| `+` | Addition | `5 + 3` | 8 |
| `-` | Subtraction | `5 - 3` | 2 |
| `*` | Multiplication | `5 * 3` | 15 |
| `/` | Division (float) | `5 / 3` | 1.6667 |
| `//` | Floor Division | `5 // 3` | 1 |
| `%` | Modulus | `5 % 3` | 2 |
| `**` | Exponentiation | `5 ** 3` | 125 |

```python
a, b = 10, 3
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)
```

## Comparison Operators

```python
x, y = 5, 3
print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= 5)
print(y <= 3)
```

## Logical Operators

```python
a, b = True, False
print(a and b)
print(a or b)
print(not a)

x, y = 10, 5
print(x > 5 and y < 10)
print(x < 5 or y < 10)
print(not(x == y))
```

## Assignment Operators

```python
x = 5
x += 3
print("After += 3:", x)
x *= 2
print("After *= 2:", x)
x //= 5
print("After //= 5:", x)
```

## Membership Operators

```python
nums = [1, 2, 5]
print(5 in nums)
print(3 in nums)
print(3 not in nums)
```

## Identity Operators

```python
a = [1,2]
b = a
c = [1,2]

print(a is b)
print(a is c)
print(a is not c)
```

## Conditional Statements
### if Statement

```python
x = 10
if x > 5:
 print("x is greater than 5")
```

### if...else Statement

```python
x = 7
if x % 2 == 0:
    print("x is even")
else:
    print("x is odd")
```

### if...elif...else Statement

```python
marks = 82
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Grade F")
```

### Nested if Statement

```python
num = 18
if num > 0:
    if num % 2 == 0:
        print("Positive even number")
    else:
        print("Positive odd number")
else:
    print("Non-positive number")
```

### Combining Conditions with Logical Operators

```python
x, y = 12, 7
if x > 10 and y < 10:
    print("x is greater than 10 AND y is less than 10")

age = 16
has_permission = True
if age >= 18 or has_permission:
    print("Allowed to enter")

flag = False
if not flag:
    print("Flag is False")
```
