### Tuple Unpacking

Extract individual elements from a tuple into separate variables.

```python
t = (1, 2, 3)
a, b, c = t
print(a, b, c)
```

```python
# tuple to list conversion
tp=(11,12,13)
lst=list(tp)
print(tp)
print(lst)
```

### Nested Tuples

Tuples can contain other tuples as elements.

```python
tp=((1,2),(3,4),(5,6))
print(tp[0])
print(tp[1][1])
```

```python
# traversal

lst=(11,12,13,14,15,16)

for item in lst:
  print(item)
```
