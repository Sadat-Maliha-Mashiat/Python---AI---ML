# Set

- unordered
- mutable : elements can be added or removed, can't modifiy a value
- **doesn't allow duplicates**
- **does not allow slicing & indexing**
- Defined using {}.

```python
# creation
st={11,21,31,41,51,11,21}
print(st)
# empty set
empst={}
print(empst)
```

```python
# Taking Int Input from user into a set
st=set(map(int,input("Enter Vslues:").split()))
```

```python
# Taking string Input from user into a set
sts=set(input("Enter Values:").split())
```

Adding Elements
- add()
- update()

```python
st={10,12,15,14}
st.add(6)
print(st)
```

```python
st={11,12,13,15}
st.update([1],{145,46})
print(st)
```

Removing
- remove()
- discard()
- pop()
- clear()

```python
s={11,21,31,41,51}
s.remove(310)
print(s)
```

```python
s={11,21,31,41,51}
s.discard(310)
print(s)
```

- in, not in

- union(), |
- intersection(), &
- difference, -
- issubset()
- issuperset()
- ==, !=
- ^, symmetric_difference()
- isdisjoint()

```python
s1={1,2,3}
s2={3,4,5}
print(s1|s2)
print(s1.union(s2))
```

```python
s1={1,2,3}
s2={3,4,5}
print(s1&s2)
print(s1.intersection(s2))
```

```python
s1={1,2,3}
s2={3,4,5}
print(s1-s2)
print(s1.difference(s2))
```

```python
s={1,2,3,4,5,6,7}
s2={2,4,5}
s3={9,10}
print(s.issubset(s2))
print(s.issubset(s3))
print(s1.issubset(s))
print(s1.issuperset(s3))
```
